```python
from csae_pyutils import gsheet_to_df
from tqdm import tqdm
from apis_core.apis_entities.models import Work, Event
from apis_core.apis_metainfo.models import Uri
from apis_core.apis_relations.models import EventWork
from apis_core.apis_vocabularies.models import EventWorkRelation
```

```python
event_work_relation = EventWorkRelation.objects.get(id=1201)
event_work_relation
```

```python
df = gsheet_to_df("1FzI7lYOKuc4LIsKKEJCUCyEzijOK4aHe7Ejm-kR0Bb0")
```

```python
for i, row in tqdm(df.iterrows(), total=len(df)):
    event = Uri.objects.get(uri=row["burgtheater_event_url"]).entity
    event = Event.objects.get(id=event.id)
    work = Uri.objects.get(uri=row["work_id"]).entity
    work = Work.objects.get(id=work.id)
    EventWork.objects.get_or_create(
        related_event=event,
        related_work=work,
        start_date_written=event.start_date_written,
        end_date_written=event.end_date_written,
        relation_type=event_work_relation
    )
    
    
```
