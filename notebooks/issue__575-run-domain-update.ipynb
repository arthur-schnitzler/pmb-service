```python
from django.db.models import Q
from apis_core.apis_metainfo.models import Uri, Collection
```

```python
Uri.objects.filter(domain="anno").filter(
    Q(uri__contains="aid=bth") | Q(uri__contains="aid=ope") | Q(uri__contains="aid=wtz")
).update(domain="anno-theaterzettel")
```

```python
Uri.objects.filter(domain="spielplan-oper_1869–1945").update(domain="spielplan-oper")
```

```python
for c in Collection.objects.filter(related_domain__contains=["spielplan-oper_1869–1945"]):
    c.related_domain = ["spielplan-oper" if d == "spielplan-oper_1869–1945" else d for d in c.related_domain]
    c.save()
```
