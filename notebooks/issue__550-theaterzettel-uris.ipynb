```python
# 2026-11-18
from django.core.exceptions import ObjectDoesNotExist
from csae_pyutils import gsheet_to_df
from tqdm import tqdm
from apis_core.apis_metainfo.models import TempEntityClass, Uri
```

```python
domain = "anno-theaterzettel"
domain
```

```python
df = gsheet_to_df("13dnBAVBzqObY3otgRuSlHVBOURfzRXX9vi7xJi9jtM0")
```

```python
df.head(3)
```

```python
for i, row in tqdm(df.iloc[7000:].iterrows(), total=len(df) - 7000):
    try:
        url = Uri.objects.get(uri=row["uri"])
    except ObjectDoesNotExist:
        print(url)
        continue
    url.domain = domain
    url.save()
```

```python
for i, row in tqdm(df.iloc[7000:].iterrows(), total=len(df) - 7000):
    pmb_uri = row["uri"]
    try:
        entity = TempEntityClass.objects.get(id=row["pmb-nummer"])
    except ObjectDoesNotExist:
        tmp_uri = f"https://pmb.acdh.oeaw.ac.at/entity/{row['pmb-nummer']}/"
        try:
            entity = Uri.objects.get(uri=tmp_uri).entity
        except ObjectDoesNotExist:
            print(tmp_uri)
            continue
    try:
        pmb_uri, _ = Uri.objects.get_or_create(uri=pmb_uri, domain=domain, entity=entity)
    except:
        pass
   
```
