```python
from csae_pyutils import gsheet_to_df
from tqdm import tqdm
from apis_core.apis_metainfo.models import normalize_wikimedia_img_url, TempEntityClass
```

```python
df = gsheet_to_df("19n07ISpkRs_Jm_CT0d6WvvtuTebKz9sZb2oB0ySfNlA")
```

```python
df.head(3)
```

```python
errors = []
for i, row in tqdm(df.iterrows(), total=len(df)):
    url = row["url"]
    normalized_url = normalize_wikimedia_img_url(url)
    try:
        entity = TempEntityClass.objects.get(id=row["pmb"])
    except Exception as e:
        errors.append([e, row["pmb"]])
        continue
    entity.img_url = normalized_url
    try:
        entity.save()
    except Exception as e:
        errors.append([e, row["pmb"]])
        continue
for x in errors:
    print(x)
```
