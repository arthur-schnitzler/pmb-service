```python
# 2026-10-19
# be aware, creates duplicates if run more than once!
from csae_pyutils import gsheet_to_df
from tqdm import tqdm
from apis_core.apis_metainfo.models import Collection, TempEntityClass
from apis_core.apis_entities.models import Place
from apis_core.apis_vocabularies.models import PlacePlaceRelation, PlaceType
from apis_core.apis_relations.models import PlacePlace
```

```python
col = Collection.objects.filter(name__startswith="Brahms Online").first()
place_place_relation = PlacePlaceRelation.objects.get(name="gehört zu")
col, place_place_relation
```

```python
df = gsheet_to_df("14ifoQ26S31Upy8KY0_eUGyhgjxrbTNrBF6PHqAwYKz8")
```

```python
df.head(3)
```

```python
for i, row in tqdm(df.iterrows(), total=len(df)):
    place = Place.objects.create(name=row["name"])
    entity = TempEntityClass.objects.get(id=place.id)
    entity.collection.add(col)
    place_type = PlaceType.objects.get(name=row["typ"])
    place.kind = place_type
    place.save()
    try:
        related_place = Place.objects.get(id=row["place-pmb"])
    except:
        print(row["name"], row["place-pmb"])
        continue
    PlacePlace.objects.create(related_placea=place, related_placeb=related_place, relation_type=place_place_relation)
```
